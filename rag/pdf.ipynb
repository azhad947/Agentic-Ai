```python
import os
from langchain_community.document_loaders import PyPDFLoader, PyMuPDFLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter
from pathlib import Path
```

```python
def process_pdf(filePath):

    pdfPath = Path(filePath)
    all_document = []
    pdf_files = list(pdfPath.glob("**/*.pdf"))
    print(f"so this is {pdf_files}")

    try:
        for pdf_file in pdf_files:
            loader = PyMuPDFLoader(file_path=pdf_file)
            document = loader.load()
            all_document.extend(document)
            print(f"{all_document}")

            
    except Exception as e:
            print(f"  ✗ Error: {e}")        
    
    print(f"\nTotal documents loaded: {len(all_document)}")
    return all_document












process_pdf("./data")
```
